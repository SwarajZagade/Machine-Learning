```python
import pandas as pd
```

```python
df=pd.read_csv('../../../datasets/Obesity Classification.csv')
```

```python
df.head()
```

```python
df['inc_gender']=df['Gender'].map({'Male' : 1,'Female': 0})
```

```python
df['inc_Label']=df['Label'].map({'Normal Weight':0,'Overweight':1,'Underweight':2,'Obese':3})
```

```python
df.head()
```

```python
df['Label'].unique()
```

```python
X=df[['Age','inc_gender','Weight']]
y=df['inc_Label']
```

```python
df.info()
```

```python
df.describe()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20)
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.preprocessing  import MinMaxScaler
scaler=MinMaxScaler()
scaler.fit(X_train)
X_train_scaled=scaler.transform(X_train)
```

```python
from sklearn.linear_model import LogisticRegression
model= LogisticRegression()
model.fit(X_train_scaled,y_train)
```

```python
X_test_scaled=scaler.transform(X_test)
model.score(X_test_scaled,y_test)
```

```python
q=[[25,1,25.3]]
model.predict(q)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
models={'LogReg':LogisticRegression(),
       'KNN-1':KNeighborsClassifier(n_neighbors=1),
       'KNN-3':KNeighborsClassifier(n_neighbors=3),
       'KNN-5':KNeighborsClassifier(n_neighbors=5),
       'KNN-7':KNeighborsClassifier(n_neighbors=7),
       'KNN-9':KNeighborsClassifier(n_neighbors=9)}

for name,model in models.items():
    model.fit(X_train_scaled,y_train)
    score = model.score(X_test_scaled,y_test)
    print(name,score)
```

```python
from sklearn.metrics import confusion_matrix,precision_score,recall_score,f1_score
from sklearn.svm import SVC
```

```python
yp=model.predict(X_test_scaled)
confusion_matrix(y_test,yp)
```

```python
models={'LogReg':LogisticRegression(),
       'KNN-1':KNeighborsClassifier(n_neighbors=1),
       'KNN-3':KNeighborsClassifier(n_neighbors=3),
       'KNN-5':KNeighborsClassifier(n_neighbors=5),
       'KNN-7':KNeighborsClassifier(n_neighbors=7),
       'KNN-9':KNeighborsClassifier(n_neighbors=9),
       'SVC-linear':SVC(kernel='linear'),
       'SVC-poly-d2':SVC(kernel='poly',degree=2),
       'SVC-poly-d3':SVC(kernel='poly',degree=3),
       'SVC-rbf':SVC(kernel='rbf')}

for name,model in models.items():
    model.fit(X_train_scaled,y_train)
    score = model.score(X_test_scaled,y_test)
    print(name,score)
    yp=model.predict(X_test_scaled)
    print("confusion_matrix \n",confusion_matrix(y_test,yp),"\n",
    "precision_score \n",precision_score(y_test,yp),"\n",
    "recall_score \n",recall_score(y_test,yp),"\n",
    "f1 score \n",f1_score(y_test,yp))
    print("\t")
    print("\t")
```
