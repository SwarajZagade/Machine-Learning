```python
import pandas as pd
```

```python
df=pd.read_csv('../../datasets/Health_insurance.csv')
df.shape
```

```python
df.head(2)
```

```python
df['sex_inc']=df['sex'].map({'female':1,'male':0})
```

```python
df.head(3)
```

```python
df['smoker_enc']=df['smoker'].map({'yes':1,'no':0})
```

```python
df.head(3)
```

```python
X=df[['age','sex_inc','bmi','children','smoker_enc']]
y=df['charges']
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(X,y)
```

```python
q=[[19,1,27.90,0,1],
  [18,0,33.77,1,0]]
model.predict(q)
```

```python
model.score(X,y)
```
