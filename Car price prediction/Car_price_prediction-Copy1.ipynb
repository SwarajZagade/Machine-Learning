```python
from sklearn.svm import SVR
```

```python
models={'LinearRegression':LinearRegression(),
       'KNN-1':KNeighborsRegressor(n_neighbors=1),
       'KNN-3':KNeighborsRegressor(n_neighbors=3),
       'KNN-5':KNeighborsRegressor(n_neighbors=5),
       'KNN-7':KNeighborsRegressor(n_neighbors=7),
       'KNN-9':KNeighborsRegressor(n_neighbors=9),
       'SVM-Linear':SVR(kernel='linear'),
       'SVM-Poly-2':SVR(kernel='poly',degree=2),
        'SVM-Poly-3':SVR(kernel='poly',degree=3),
       'SVM-rbf':SVR(kernel='rbf')}

for name,model in models.items():
    model.fit(X_train_scaled,y_train)
    score = model.score(X_test_scaled,y_test)
    print(name,score)
```
