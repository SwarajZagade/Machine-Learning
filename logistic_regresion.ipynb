```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('../../datasets/Social_Network_Ads.csv')
df.shape
```

```python
df.head(4)
```

```python
plt.scatter(df['Age'],df['EstimatedSalary'],c=df['Purchased'],cmap='RdYlBu')
```

```python
X=df[['Age','EstimatedSalary']]
y=df['Purchased']
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X,y)
```

```python
q=[[50,120000],
    [30,40000]]
model.predict(q)
```

```python
yp=model.predict(X)
sum(yp==y)/400
```

```python
model.score(X,y) # accuracy
```
